```python
import scanpy as sc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

sc.settings.verbosity = 3
sc.logging.print_versions()

import gc
```

```python
gc.collect()
```

```python
adata = sc.read_10x_mtx("/h/am/EX_M/datasets/scRNA/pbmc68k/data/pbmc68k/filtered_matrices_mex/hg19", var_names='gene_symbols', cache=False)                              
```

```python
adata.var_names_make_unique()
sc.pp.filter_genes(adata, min_cells=3)
# sc.pp.filter_cells(adata, min_genes=10) # already done in the original dataset
adata.layers["counts"] = adata.X.copy()
```

```python
sc.pp.normalize_total(adata, target_sum=1e4)
sc.pp.log1p(adata)

celltype = pd.read_csv('/h/am/EX_M/datasets/scRNA/pbmc68k/data/pbmc68k/filtered_matrices_mex/68k_pbmc_barcodes_annotation.tsv', sep='\t')['celltype'].values

adata.obs['celltype'] = celltype
```

```python
adata.write('/h/am/EX_M/datasets/scRNA/pbmc68k/pbmc68k.h5ad')
```
