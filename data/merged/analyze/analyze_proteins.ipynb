```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib import rcParams
from matplotlib import colors
import seaborn as sns
import gc

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv('../../../../EX_M/datasets/scRNA/merged/gene_to_protein.csv', index_col=0)
df
```

```python
df['gene_biotype'].value_counts()
```

```python
sorted_df = df.sort_values(by='n_cells')
sorted_df
```

```python
sorted_df.head(100).plot(kind='bar', figsize=(28, 5), title='All Genes', x='gene_symbol', y='n_cells', color='blue');
```

```python
sorted_df.tail(100).plot(kind='bar', figsize=(28, 5), title='All Genes', x='gene_symbol', y='n_cells', color='blue');
```

```python
sorted_df[sorted_df['gene_biotype'] == "protein_coding"].head(100).plot(kind='bar', figsize=(28, 5), title='Protein Coding Genes', x='gene_symbol', y='n_cells', color='blue');
```

```python
sorted_df[sorted_df['gene_biotype'] == "protein_coding"].tail(100).plot(kind='bar', figsize=(28, 5), title='Protein Coding Genes', x='gene_symbol', y='n_cells', color='blue');
```

```python
sorted_df[sorted_df['gene_biotype'] != "protein_coding"].head(100).plot(kind='bar', figsize=(28, 5), title='Other Genes', x='gene_symbol', y='n_cells', color='blue');
```
